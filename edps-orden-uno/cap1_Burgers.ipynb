# Ecuación de Burgers: formación de un shock, 1D

Traducción de `cap1_Burgers.m`.

$$u_t + \left(\tfrac{u^2}{2}\right)_x = 0$$

Con el dato inicial tipo rampa decreciente, las características se cruzan en
$t=1$: hasta ese momento la solución es continua (rampa que se va parando) y
después aparece un shock que viaja con velocidad $1/2$ (Rankine–Hugoniot), es
decir ubicado en $x=(1+t)/2$.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import animation
from IPython.display import HTML
```

## Solución exacta

Versión vectorizada de la función `sol` del script (así no hace falta el bucle
interno sobre `xx`).

```python
def sol(x, t):
    x = np.asarray(x, dtype=float)
    if t < 1:
        # rampa: 1 detras del frente, decreciente, 0 adelante
        return np.where(x <= t, 1.0,
               np.where(x <= 1, 1.0 - (x - t)/(1.0 - t), 0.0))
    else:
        # shock en x = (1+t)/2
        return np.where(x > (1.0 + t)/2, 0.0, 1.0)
```

## Animación

```python
tt = np.linspace(0, 4, 100)              # linspace(0,4) en MATLAB = 100 puntos
xx = np.linspace(0, 4, 100)

fig, ax = plt.subplots(figsize=(7, 4))
ax.set_xlim(0, 4)
ax.set_ylim(0, 1)
ax.set_xlabel('x')
ax.set_ylabel('u')
linea, = ax.plot([], [])

def actualizar(i):
    t = tt[i]
    linea.set_data(xx, sol(xx, t))
    ax.set_title(f't = {t:.2f}')
    return (linea,)

ani = animation.FuncAnimation(fig, actualizar, frames=len(tt),
                              interval=120, blit=False)
plt.close(fig)
HTML(ani.to_jshtml())
```

## Algunos instantes en un solo gráfico

Útil para ver el mecanismo sin depender de la animación.

```python
fig, ax = plt.subplots(figsize=(7, 4))
for t in [0, 0.5, 0.9, 1.0, 2.0, 3.0]:
    ax.plot(xx, sol(xx, t), label=f't = {t:g}')
ax.set_xlim(0, 4)
ax.set_ylim(-0.05, 1.05)
ax.set_xlabel('x')
ax.set_ylabel('u')
ax.legend()
plt.show()
```
