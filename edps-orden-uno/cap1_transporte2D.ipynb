# Ecuación de transporte, 2D

Traducción de `cap1_transporte2D.m`.

$$u(x,y,t)=\sin\!\left(\sqrt{(x-b_1 t)^2+(y-b_2 t)^2}\right)$$

Onda circular cuyo centro se traslada con velocidad $b=(b_1,b_2)$.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import animation
from IPython.display import HTML
```

## Parámetros

**Nota:** el script original usa una malla de `1e3 x 1e3` puntos. En matplotlib
dibujar una superficie de un millón de puntos por cuadro es demasiado lento, así
que aquí se usan 200 puntos por dirección (visualmente equivalente). Subilo si
querés más detalle y tenés paciencia.

```python
b = np.array([10, 2])                    # velocidad
u = lambda x, y, t: np.sin(np.sqrt((x - b[0]*t)**2 + (y - b[1]*t)**2))

N  = 200                                 # original: 1e3
xx = np.linspace(-20, 20, N)
tt = np.linspace(0, 15, 150)
x, y = np.meshgrid(xx, xx)
```

## Animación de la superficie

```python
fig = plt.figure(figsize=(7, 5.5))
ax = fig.add_subplot(projection='3d')

frames = range(0, len(tt), 2)

def actualizar(j):
    ax.clear()                           # hay que redibujar la superficie completa
    ax.plot_surface(x, y, u(x, y, tt[j]), cmap='viridis',
                    linewidth=0, antialiased=False, rstride=2, cstride=2)
    ax.set_xlim(-20, 20)
    ax.set_ylim(-20, 20)
    ax.set_zlim(-1, 1)
    ax.set_xlabel('x')
    ax.set_ylabel('y')
    ax.set_title(f'{tt[j]:g}')

ani = animation.FuncAnimation(fig, actualizar, frames=frames,
                              interval=50, blit=False)   # FrameRate = 20 -> 50 ms
plt.close(fig)
HTML(ani.to_jshtml())
```

## Guardar el video

Equivale a `VideoWriter('transporte.avi')`. Colab ya trae `ffmpeg`, así que se
puede guardar directamente en mp4 (más liviano que avi). Si por alguna razón no
estuviera disponible, se cae de vuelta a un GIF.

```python
import shutil

if shutil.which('ffmpeg'):
    archivo = 'transporte.mp4'
    ani.save(archivo, writer='ffmpeg', fps=20)
else:                                    # sin ffmpeg: GIF con Pillow
    archivo = 'transporte.gif'
    ani.save(archivo, writer='pillow', fps=20)

print('listo:', archivo)
```
