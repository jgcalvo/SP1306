# Ecuación de transporte, 1D

Traducción de `cap1_transporte1D.m`.

$$u_t + b\,u_x = 0,\qquad u(x,t)=\sin(x-bt)$$

La solución es una onda que viaja hacia la derecha con velocidad $b$.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import animation
from IPython.display import HTML
```

## Parámetros y solución exacta

```python
b = 0.1                                  # velocidad
u = lambda x, t: np.sin(x - b*t)         # solucion exacta

xx = np.linspace(-10, 10, 100)           # linspace(a,b) en MATLAB = 100 puntos
tt = np.linspace(0, 10, 100)
```

## Gráfico de superficie (`surf`)

```python
x, t = np.meshgrid(xx, tt)               # equivalente a meshgrid de MATLAB

fig = plt.figure(figsize=(7, 5))
ax = fig.add_subplot(projection='3d')
ax.plot_surface(x, t, u(x, t), cmap='viridis', linewidth=0, antialiased=False)
ax.set_xlabel('x')
ax.set_ylabel('t')
ax.set_zlabel('u')
plt.show()
```

## Animación 1D

En MATLAB el bucle usa `pause(.1)` para ver el cambio en el tiempo. En Colab eso no
funciona: se arma una animación con `FuncAnimation` y se muestra como reproductor
interactivo con `to_jshtml()` (no necesita ffmpeg).

```python
fig, ax = plt.subplots(figsize=(7, 4))
ax.set_xlim(-10, 10)
ax.set_ylim(-1, 1)
ax.set_xlabel('x')
linea, = ax.plot([], [])

frames = range(0, len(tt), 5)            # se grafican algunas curvas

def actualizar(j):
    linea.set_data(xx, u(xx, tt[j]))
    ax.set_title(f'{tt[j]:g}')
    return (linea,)

ani = animation.FuncAnimation(fig, actualizar, frames=frames,
                              interval=100, blit=False)
plt.close(fig)                           # evita mostrar el cuadro estatico extra
HTML(ani.to_jshtml())
```
